```python
import keras
from keras.applications.resnet50 import ResNet50
from keras.applications.resnet50 import preprocess_input, decode_predictions
import numpy as np

from tensorflow.keras.preprocessing import image
```

```python
model=ResNet50(weights='imagenet')
```

```python
img_path = '/content/cat image.jpg'

# 1. Image Load & Resize (ResNet/VGG standard size: 224x224)
img = image.load_img(img_path, target_size=(224, 224))

# 2. Image ko Array me Convert Karein
img_array = image.img_to_array(img)

# 3. Reshape / Expand Dimensions (3D -> 4D Batch Shape: (1, 224, 224, 3))
x = np.expand_dims(img_array, axis=0)
# Ya fir aapka reshape method: x = img_array.reshape(1, 224, 224, 3)

# 4. Preprocess Input (Model specific scaling & color channel formatting)
x = preprocess_input(x)

print("Preprocessed shape:", x.shape)
```

```python
pred=model.predict(x)
print("Predicted:", decode_predictions(pred, top=3)[0])
```
